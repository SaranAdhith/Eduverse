# 03 — BKT predictive accuracy

Leave-last-out: for each response the model's **prior** gives a predicted P(correct) = `prior*(1-slip) + (1-prior)*guess`. Compute AUC and Brier against observed correctness, per topic and overall. This grounds the "BKT is doing real work" claim.

```python
import os
import sys

sys.path.insert(0, os.path.abspath('..'))
from lib import load, metrics, plots

EXPORT_PATH = os.environ.get('EXPORT_PATH', '../fixtures/P001.tar.gz')
plots.apply_style()
```

## Predicted vs observed


```python
pe = next(iter(load.read_tarball(EXPORT_PATH).participants.values()))
responses = pe.tables['responses']
# slip/guess come from data/bkt_params.yaml; use defaults here.
SLIP, GUESS = 0.1, 0.2
predicted = [metrics.predicted_correct(float(r['prior_mastery']), SLIP, GUESS) for r in responses]
observed = [1 if r['is_correct'] in ('True','true','1') else 0 for r in responses]
{'auc': metrics.auc(predicted, observed), 'brier': metrics.brier_score(predicted, observed)}
```
