## Load the export


```python
export = load.read_tarball(EXPORT_PATH)
df = load.load_dataframes(EXPORT_PATH) if len(export.participants)==1 else None
list(export.participants)
```

## Cohort summary
N enrolled, dropout, items answered, on-task time.

```python
rows = []
for code, pe in export.participants.items():
    responses = pe.tables.get('responses', [])
    steps = pe.tables.get('path_steps', [])
    rows.append({
        'code': code,
        'block_order': pe.participant.get('block_order'),
        'items_answered': len(responses),
        'steps': len(steps),
        'passed': sum(1 for s in steps if s['status']=='passed'),
    })
rows
```
